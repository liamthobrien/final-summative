```python
# Generate Sample Data
import pandas as pd
import numpy as np
import os
```

```python
# Simple sample data
countries = ['Afghanistan', 'India', 'China', 'USA', 'France']
years = list(range(2000, 2024))
```

```python
# Create sample yields
data = []
for country in countries:
    for year in years:
        yield_value = 2000 + (year - 2000) * 50 + np.random.normal(0, 200)
        data.append({'Country': country, 'Year': year, 'Yield': yield_value})
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.to_csv('crop_yield_data.csv', index=False)
```

```python
# Test Visualisation

import matplotlib.pyplot as plt

# Create a simple plot
fig, ax = plt.subplots(figsize=(10, 6))

for country in ['Afghanistan', 'India', 'China']:
    country_data = df[df['Country'] == country]
    ax.plot(country_data['Year'], country_data['Yield'], 
            marker='o', label=country, linewidth=2)

ax.set_title("Wheat Yield Over Time", fontsize=16)
ax.set_xlabel("Year")
ax.set_ylabel("Yield (kg/ha)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
